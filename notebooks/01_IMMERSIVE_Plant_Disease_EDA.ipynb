```python
# 🗂️ INTELLIGENT DATA INGESTION PIPELINE

class PlantDiseaseDataLoader:
    """Advanced data loading system with metadata extraction and validation"""
    
    def __init__(self, data_dir: str):
        self.data_dir = Path(data_dir)
        self.metadata = []
        self.class_mapping = {}
        self.species_mapping = {}
        
    def extract_class_info(self, folder_name: str) -> dict:
        """Extract species and disease information from folder name"""
        parts = folder_name.split('___')
        
        if len(parts) >= 2:
            species = parts[0].replace('_', ' ').title()
            condition = parts[1].replace('_', ' ').title()
        else:
            species = folder_name.split('_')[0].title()
            condition = 'Unknown'
            
        return {
            'species': species,
            'condition': condition,
            'is_healthy': 'healthy' in condition.lower(),
            'severity': self._assess_severity(condition)
        }
    
    def _assess_severity(self, condition: str) -> str:
        """Assess disease severity from condition name"""
        condition_lower = condition.lower()
        if 'healthy' in condition_lower:
            return 'None'
        elif any(word in condition_lower for word in ['early', 'mild', 'light']):
            return 'Mild'
        elif any(word in condition_lower for word in ['late', 'severe', 'advanced']):
            return 'Severe'
        else:
            return 'Moderate'
    
    def extract_image_metadata(self, image_path: Path) -> dict:
        """Extract comprehensive metadata from image file"""
        try:
            with Image.open(image_path) as img:
                # Basic properties
                width, height = img.size
                mode = img.mode
                format_type = img.format
                
                # File properties
                file_size = image_path.stat().st_size
                
                # Convert to RGB if needed for analysis
                if mode != 'RGB':
                    img = img.convert('RGB')
                
                # Color statistics
                img_array = np.array(img)
                
                return {
                    'width': width,
                    'height': height,
                    'aspect_ratio': width / height,
                    'total_pixels': width * height,
                    'mode': mode,
                    'format': format_type,
                    'file_size_mb': file_size / (1024 * 1024),
                    'mean_red': np.mean(img_array[:, :, 0]),
                    'mean_green': np.mean(img_array[:, :, 1]),
                    'mean_blue': np.mean(img_array[:, :, 2]),
                    'std_red': np.std(img_array[:, :, 0]),
                    'std_green': np.std(img_array[:, :, 1]),
                    'std_blue': np.std(img_array[:, :, 2]),
                    'brightness': np.mean(img_array),
                    'contrast': np.std(img_array)
                }
        except Exception as e:
            print(f"Error processing {image_path}: {e}")
            return None
    
    def load_dataset(self, sample_size: int = None) -> pd.DataFrame:
        """Load dataset with comprehensive metadata extraction"""
        print("🔄 Starting intelligent data ingestion...")
        
        # Discover all class directories
        class_dirs = [d for d in self.data_dir.iterdir() if d.is_dir()]
        print(f"📁 Found {len(class_dirs)} disease classes")
        
        total_images = 0
        for class_dir in class_dirs:
            image_files = list(class_dir.glob('*.jpg')) + list(class_dir.glob('*.jpeg')) + list(class_dir.glob('*.png'))
            total_images += len(image_files)
        
        print(f"🖼️ Total images discovered: {total_images:,}")
        
        if sample_size and sample_size < total_images:
            print(f"📊 Sampling {sample_size:,} images for analysis")
        
        # Process each class
        processed_count = 0
        
        for class_dir in tqdm(class_dirs, desc="Processing classes"):
            class_name = class_dir.name
            class_info = self.extract_class_info(class_name)
            
            # Find all image files
            image_extensions = ['*.jpg', '*.jpeg', '*.png', '*.bmp']
            image_files = []
            for ext in image_extensions:
                image_files.extend(class_dir.glob(ext))
            
            # Sample if requested
            if sample_size:
                images_per_class = min(len(image_files), sample_size // len(class_dirs) + 10)
                image_files = np.random.choice(image_files, min(images_per_class, len(image_files)), replace=False)
            
            # Process each image
            for img_path in image_files:
                if sample_size and processed_count >= sample_size:
                    break
                    
                img_metadata = self.extract_image_metadata(img_path)
                
                if img_metadata:
                    record = {
                        'image_path': str(img_path),
                        'filename': img_path.name,
                        'class_name': class_name,
                        'species': class_info['species'],
                        'condition': class_info['condition'],
                        'is_healthy': class_info['is_healthy'],
                        'severity': class_info['severity'],
                        **img_metadata
                    }
                    
                    self.metadata.append(record)
                    processed_count += 1
            
            if sample_size and processed_count >= sample_size:
                break
        
        df = pd.DataFrame(self.metadata)
        print(f"✅ Dataset loaded successfully!")
        print(f"📊 Final dataset: {len(df):,} images across {df['class_name'].nunique()} classes")
        print(f"🌱 Species: {df['species'].nunique()} ({', '.join(df['species'].unique())})")
        print(f"🏥 Conditions: {df['condition'].nunique()} disease states")
        
        return df

# Initialize the data loader
data_dir = "../data"
loader = PlantDiseaseDataLoader(data_dir)

# Load a comprehensive sample for analysis (adjust size based on computational resources)
SAMPLE_SIZE = 2000  # Set to None for full dataset
df = loader.load_dataset(sample_size=SAMPLE_SIZE)

print(f"\n🎯 DATASET READY FOR IMMERSIVE ANALYSIS")
print(f"📈 Shape: {df.shape}")
print(f"💾 Memory usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```

```python
# 🎨 COMPREHENSIVE COLOR ANALYSIS - Multi-Library Visualization Suite

# Prepare color data
color_data = df[['mean_red', 'mean_green', 'mean_blue', 'std_red', 'std_green', 'std_blue', 
                'brightness', 'contrast', 'species', 'is_healthy', 'condition']].copy()

# Calculate additional color metrics
color_data['rgb_sum'] = color_data['mean_red'] + color_data['mean_green'] + color_data['mean_blue']
color_data['green_dominance'] = color_data['mean_green'] / (color_data['rgb_sum'] + 1e-8)
color_data['red_green_ratio'] = color_data['mean_red'] / (color_data['mean_green'] + 1e-8)
color_data['color_variance'] = (color_data['std_red'] + color_data['std_green'] + color_data['std_blue']) / 3

# Create subplot structure
fig = make_subplots(
    rows=3, cols=2,
    subplot_titles=(
        'RGB Channel Distributions', 'Color Space 3D Scatter',
        'Brightness vs Contrast by Health', 'Green Dominance by Species',
        'Color Variance Analysis', 'RGB Correlation Heatmap'
    ),
    specs=[[{"secondary_y": False}, {"type": "scatter3d"}],
           [{"secondary_y": False}, {"secondary_y": False}],
           [{"secondary_y": False}, {"type": "heatmap"}]],
    vertical_spacing=0.08,
    horizontal_spacing=0.1
)

# 1. RGB Channel Distributions (Violin plots)
for i, channel in enumerate(['mean_red', 'mean_green', 'mean_blue']):
    fig.add_trace(
        go.Violin(
            y=color_data[channel],
            name=channel.split('_')[1].upper(),
            box_visible=True,
            meanline_visible=True,
            fillcolor=['red', 'green', 'blue'][i],
            opacity=0.6,
            line_color='black'
        ),
        row=1, col=1
    )

# 2. 3D Color Space Scatter
health_colors = ['red' if not h else 'green' for h in color_data['is_healthy']]
fig.add_trace(
    go.Scatter3d(
        x=color_data['mean_red'],
        y=color_data['mean_green'], 
        z=color_data['mean_blue'],
        mode='markers',
        marker=dict(
            color=health_colors,
            size=4,
            opacity=0.7
        ),
        text=color_data['condition'],
        name='RGB Space',
        showlegend=False
    ),
    row=1, col=2
)

# 3. Brightness vs Contrast by Health Status
for health_status in [True, False]:
    subset = color_data[color_data['is_healthy'] == health_status]
    fig.add_trace(
        go.Scatter(
            x=subset['brightness'],
            y=subset['contrast'],
            mode='markers',
            marker=dict(
                color='green' if health_status else 'red',
                size=6,
                opacity=0.6
            ),
            name='Healthy' if health_status else 'Diseased',
            text=subset['species']
        ),
        row=2, col=1
    )

# 4. Green Dominance by Species (Box plots)
for i, species in enumerate(color_data['species'].unique()):
    species_data = color_data[color_data['species'] == species]
    fig.add_trace(
        go.Box(
            y=species_data['green_dominance'],
            name=species,
            boxmean='sd',
            fillcolor=px.colors.qualitative.Set1[i % len(px.colors.qualitative.Set1)],
            opacity=0.7
        ),
        row=2, col=2
    )

# 5. Color Variance by Condition (Strip plot style)
conditions = color_data['condition'].value_counts().head(8).index  # Top 8 conditions
for i, condition in enumerate(conditions):
    condition_data = color_data[color_data['condition'] == condition]
    fig.add_trace(
        go.Scatter(
            x=[condition] * len(condition_data),
            y=condition_data['color_variance'],
            mode='markers',
            marker=dict(
                color=px.colors.qualitative.Pastel[i % len(px.colors.qualitative.Pastel)],
                size=5,
                opacity=0.6
            ),
            name=condition[:15],  # Truncate long names
            showlegend=False
        ),
        row=3, col=1
    )

# 6. RGB Correlation Heatmap
rgb_corr = color_data[['mean_red', 'mean_green', 'mean_blue', 'brightness', 'contrast']].corr()
fig.add_trace(
    go.Heatmap(
        z=rgb_corr.values,
        x=rgb_corr.columns,
        y=rgb_corr.index,
        colorscale='RdBu',
        zmid=0,
        text=rgb_corr.values.round(3),
        texttemplate="%{text}",
        showscale=False
    ),
    row=3, col=2
)

# Update layout
fig.update_layout(
    height=1200,
    width=1200,
    title={
        'text': "🎨 Comprehensive Color Analysis Dashboard",
        'x': 0.5,
        'font': {'size': 20}
    },
    showlegend=True
)

# Update axes labels
fig.update_xaxes(title_text="RGB Channels", row=1, col=1)
fig.update_yaxes(title_text="Pixel Intensity", row=1, col=1)
fig.update_xaxes(title_text="Brightness", row=2, col=1)
fig.update_yaxes(title_text="Contrast", row=2, col=1)
fig.update_xaxes(title_text="Species", row=2, col=2)
fig.update_yaxes(title_text="Green Dominance", row=2, col=2)
fig.update_xaxes(title_text="Disease Condition", row=3, col=1, tickangle=45)
fig.update_yaxes(title_text="Color Variance", row=3, col=1)

fig.show()

# Advanced color statistics
print("🎨 ADVANCED COLOR INSIGHTS:")
print(f"\n🔴 Red Channel Analysis:")
print(f"   • Healthy plants: {color_data[color_data['is_healthy']]['mean_red'].mean():.2f} ± {color_data[color_data['is_healthy']]['mean_red'].std():.2f}")
print(f"   • Diseased plants: {color_data[~color_data['is_healthy']]['mean_red'].mean():.2f} ± {color_data[~color_data['is_healthy']]['mean_red'].std():.2f}")

print(f"\n🟢 Green Channel Analysis:")
print(f"   • Healthy plants: {color_data[color_data['is_healthy']]['mean_green'].mean():.2f} ± {color_data[color_data['is_healthy']]['mean_green'].std():.2f}")
print(f"   • Diseased plants: {color_data[~color_data['is_healthy']]['mean_green'].mean():.2f} ± {color_data[~color_data['is_healthy']]['mean_green'].std():.2f}")

print(f"\n💡 Brightness & Contrast:")
healthy_brightness = color_data[color_data['is_healthy']]['brightness'].mean()
diseased_brightness = color_data[~color_data['is_healthy']]['brightness'].mean()
print(f"   • Brightness difference: {abs(healthy_brightness - diseased_brightness):.2f} (healthy vs diseased)")

print(f"\n🌿 Green Dominance Analysis:")
species_green_dom = color_data.groupby('species')['green_dominance'].agg(['mean', 'std'])
for species, stats in species_green_dom.iterrows():
    print(f"   • {species}: {stats['mean']:.3f} ± {stats['std']:.3f}")

# Statistical test for color differences
from scipy.stats import ttest_ind
healthy_green = color_data[color_data['is_healthy']]['mean_green']
diseased_green = color_data[~color_data['is_healthy']]['mean_green']
t_stat, p_value = ttest_ind(healthy_green, diseased_green)
print(f"\n📊 Statistical Test (Green Channel):")
print(f"   • t-statistic: {t_stat:.3f}")
print(f"   • p-value: {p_value:.6f}")
print(f"   • Significant difference: {'Yes' if p_value < 0.05 else 'No'}")
```
