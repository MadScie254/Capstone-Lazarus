```python
# 🏗️ ADVANCED DATA INGESTION & METADATA EXTRACTION ENGINE

class PlantDiseaseAnalyzer:
    """Ultimate plant disease data analysis system with comprehensive metadata extraction"""
    
    def __init__(self, data_dir: str):
        self.data_dir = Path(data_dir)
        self.metadata = []
        self.color_profiles = {}
        self.texture_features = {}
        self.morphological_data = {}
        
        # Disease severity mapping
        self.severity_mapping = {
            'healthy': 0,
            'early': 1, 'mild': 1, 'light': 1,
            'moderate': 2, 'medium': 2,
            'late': 3, 'severe': 3, 'advanced': 3
        }
        
        # Pathogen type mapping
        self.pathogen_mapping = {
            'blight': 'fungal',
            'rust': 'fungal', 
            'spot': 'fungal',
            'mold': 'fungal',
            'bacterial': 'bacterial',
            'virus': 'viral',
            'mosaic': 'viral',
            'curl': 'viral',
            'mite': 'pest'
        }
    
    def extract_disease_intelligence(self, folder_name: str) -> dict:
        """Extract comprehensive disease information with AI-enhanced parsing"""
        parts = folder_name.replace('___', '_').replace('__', '_').split('_')
        
        # Extract species
        species_candidates = ['corn', 'maize', 'tomato', 'potato']
        species = 'unknown'
        for part in parts:
            if any(s in part.lower() for s in species_candidates):
                species = part.lower().replace('(', '').replace(')', '')
                break
        
        # Extract condition
        condition_parts = [p for p in parts if p.lower() not in [species, 'maize']]
        condition = ' '.join(condition_parts) if condition_parts else 'unknown'
        
        # Determine health status
        is_healthy = 'healthy' in condition.lower()
        
        # Assess severity
        severity_score = 0
        for term, score in self.severity_mapping.items():
            if term in condition.lower():
                severity_score = max(severity_score, score)
        
        # Determine pathogen type
        pathogen_type = 'unknown'
        for term, ptype in self.pathogen_mapping.items():
            if term in condition.lower():
                pathogen_type = ptype
                break
        
        return {
            'species': species.title(),
            'condition': condition.title(),
            'is_healthy': is_healthy,
            'severity_score': severity_score,
            'severity_label': ['Healthy', 'Mild', 'Moderate', 'Severe'][severity_score],
            'pathogen_type': pathogen_type,
            'economic_risk': ['None', 'Low', 'Medium', 'High'][severity_score]
        }
    
    def extract_advanced_image_features(self, image_path: Path) -> dict:
        """Extract comprehensive image features for deep analysis"""
        try:
            # Load image
            with Image.open(image_path) as pil_img:
                # Basic properties
                width, height = pil_img.size
                mode = pil_img.mode
                
                # Convert to numpy array
                if mode != 'RGB':
                    pil_img = pil_img.convert('RGB')
                
                img_array = np.array(pil_img)
                
                # Basic statistics
                file_size = image_path.stat().st_size
                
                # Color space analysis
                rgb_mean = np.mean(img_array, axis=(0, 1))
                rgb_std = np.std(img_array, axis=(0, 1))
                
                # Convert to different color spaces for analysis
                hsv_img = cv2.cvtColor(img_array, cv2.COLOR_RGB2HSV)
                lab_img = cv2.cvtColor(img_array, cv2.COLOR_RGB2LAB)
                
                # HSV statistics
                hsv_mean = np.mean(hsv_img, axis=(0, 1))
                hsv_std = np.std(hsv_img, axis=(0, 1))
                
                # LAB statistics  
                lab_mean = np.mean(lab_img, axis=(0, 1))
                lab_std = np.std(lab_img, axis=(0, 1))
                
                # Texture features using gray image
                gray_img = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)
                
                # Edge detection features
                edges_sobel = cv2.Sobel(gray_img, cv2.CV_64F, 1, 1, ksize=3)
                edge_density = np.mean(np.abs(edges_sobel))
                
                # Texture analysis using Local Binary Patterns
                from skimage.feature import local_binary_pattern
                lbp = local_binary_pattern(gray_img, P=8, R=1, method='uniform')
                lbp_hist, _ = np.histogram(lbp.ravel(), bins=10)
                lbp_variance = np.var(lbp_hist)
                
                # Color diversity metrics
                unique_colors = len(np.unique(img_array.reshape(-1, 3), axis=0))
                color_diversity = unique_colors / (width * height)
                
                # Green vegetation indices
                green_ratio = rgb_mean[1] / (np.sum(rgb_mean) + 1e-8)
                excess_green = 2 * rgb_mean[1] - rgb_mean[0] - rgb_mean[2]
                
                # Brightness and contrast metrics
                brightness = np.mean(img_array)
                contrast = np.std(img_array)
                
                # Image quality metrics
                sharpness = cv2.Laplacian(gray_img, cv2.CV_64F).var()
                
                return {
                    # Basic properties
                    'width': width,
                    'height': height,
                    'aspect_ratio': width / height,
                    'total_pixels': width * height,
                    'file_size_mb': file_size / (1024 * 1024),
                    'compression_ratio': (width * height * 3) / file_size,
                    
                    # RGB statistics
                    'mean_red': rgb_mean[0],
                    'mean_green': rgb_mean[1], 
                    'mean_blue': rgb_mean[2],
                    'std_red': rgb_std[0],
                    'std_green': rgb_std[1],
                    'std_blue': rgb_std[2],
                    
                    # HSV statistics
                    'mean_hue': hsv_mean[0],
                    'mean_saturation': hsv_mean[1],
                    'mean_value': hsv_mean[2],
                    'std_hue': hsv_std[0],
                    'std_saturation': hsv_std[1],
                    'std_value': hsv_std[2],
                    
                    # LAB statistics
                    'mean_l': lab_mean[0],
                    'mean_a': lab_mean[1],
                    'mean_b': lab_mean[2],
                    'std_l': lab_std[0],
                    'std_a': lab_std[1],
                    'std_b': lab_std[2],
                    
                    # Advanced color metrics
                    'brightness': brightness,
                    'contrast': contrast,
                    'green_ratio': green_ratio,
                    'excess_green': excess_green,
                    'color_diversity': color_diversity,
                    
                    # Texture and quality metrics
                    'edge_density': edge_density,
                    'lbp_variance': lbp_variance,
                    'sharpness': sharpness,
                    'unique_colors': unique_colors
                }
                
        except Exception as e:
            print(f"Error processing {image_path}: {e}")
            return None
    
    def load_comprehensive_dataset(self, sample_size: int = None, include_features: bool = True) -> pd.DataFrame:
        """Load dataset with comprehensive feature extraction"""
        
        print("🚀 INITIATING COMPREHENSIVE DATA INGESTION")
        print("=" * 60)
        
        # Discover dataset structure
        class_dirs = [d for d in self.data_dir.iterdir() if d.is_dir()]
        print(f"📁 Discovered {len(class_dirs)} disease classes")
        
        # Count total images
        total_images = 0
        for class_dir in class_dirs:
            extensions = ['*.jpg', '*.jpeg', '*.png', '*.bmp']
            for ext in extensions:
                total_images += len(list(class_dir.glob(ext)))
        
        print(f"🖼️ Total images discovered: {total_images:,}")
        
        if sample_size and sample_size < total_images:
            print(f"📊 Strategic sampling: {sample_size:,} images for analysis")
        
        # Process each class with progress tracking
        processed_count = 0
        start_time = time.time()
        
        for class_dir in tqdm(class_dirs, desc="🔬 Processing disease classes"):
            class_name = class_dir.name
            disease_info = self.extract_disease_intelligence(class_name)
            
            # Find all image files
            image_extensions = ['*.jpg', '*.jpeg', '*.png', '*.bmp']
            image_files = []
            for ext in image_extensions:
                image_files.extend(class_dir.glob(ext))
            
            # Strategic sampling per class
            if sample_size:
                images_per_class = min(len(image_files), sample_size // len(class_dirs) + 10)
                if len(image_files) > images_per_class:
                    image_files = np.random.choice(image_files, images_per_class, replace=False)
            
            # Process images in this class
            class_start = time.time()
            
            for img_path in image_files:
                if sample_size and processed_count >= sample_size:
                    break
                
                if include_features:
                    img_features = self.extract_advanced_image_features(img_path)
                else:
                    img_features = {
                        'width': 224, 'height': 224, 'aspect_ratio': 1.0,
                        'total_pixels': 50176, 'file_size_mb': 0.1
                    }
                
                if img_features:
                    record = {
                        'image_path': str(img_path),
                        'filename': img_path.name,
                        'class_name': class_name,
                        **disease_info,
                        **img_features,
                        'processing_time': time.time() - class_start
                    }
                    
                    self.metadata.append(record)
                    processed_count += 1
            
            if sample_size and processed_count >= sample_size:
                break
        
        # Create comprehensive dataframe
        df = pd.DataFrame(self.metadata)
        total_time = time.time() - start_time
        
        print(f"\n✅ DATASET INGESTION COMPLETE")
        print("=" * 60)
        print(f"📊 Final dataset: {len(df):,} samples × {len(df.columns)} features")
        print(f"🌱 Species: {df['species'].nunique()} ({', '.join(df['species'].unique())})")
        print(f"🏥 Conditions: {df['condition'].nunique()} disease states")
        print(f"🦠 Pathogen types: {df['pathogen_type'].nunique()} ({', '.join(df['pathogen_type'].unique())})")
        print(f"⚡ Processing rate: {processed_count/total_time:.1f} images/second")
        print(f"💾 Dataset size: {df.memory_usage(deep=True).sum()/(1024**2):.2f} MB")
        
        return df

# Initialize the advanced analyzer
print("🏗️ Initializing Plant Disease Analyzer...")
data_dir = "../data"
analyzer = PlantDiseaseAnalyzer(data_dir)

# Load comprehensive dataset (adjust sample_size based on computational resources)
SAMPLE_SIZE = 3000  # Set to None for full dataset analysis
INCLUDE_ADVANCED_FEATURES = True  # Set to False for faster loading

df = analyzer.load_comprehensive_dataset(
    sample_size=SAMPLE_SIZE, 
    include_features=INCLUDE_ADVANCED_FEATURES
)

print(f"\n🎯 ADVANCED DATASET READY FOR IMMERSIVE ANALYSIS")
print(f"🔬 Enhanced with {len([c for c in df.columns if 'mean_' in c or 'std_' in c])} color space features")
print(f"🖼️ Texture analysis: {len([c for c in df.columns if any(t in c for t in ['edge', 'lbp', 'sharp'])])} features")
print(f"📊 Ready for 200+ visualization cells!")
```
